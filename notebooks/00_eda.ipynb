```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "configs/experiment.toml").exists())
sys.path.insert(0, str(ROOT / "src"))
from heart_disease.config import load_settings
from heart_disease.data import load_cleveland, split_dataset
settings = load_settings()
raw = load_cleveland(settings.raw_path)
splits = split_dataset(raw, settings)
```

```python
print(raw.shape)
display(raw.head())
from heart_disease.eda import summarize
summary = summarize(raw)
print(summary["overview"])
display(summary["dtypes"], summary["missing"], summary["target_counts"], summary["describe"])
```

## Chia tập và vẽ trên train

Những quyết định dùng nhãn để chọn đặc trưng chỉ học từ train. Giữ index để đối chiếu cùng một bệnh nhân giữa các biến thể.

```python
for name, (X, y) in splits.items():
    print(name, X.shape, y.value_counts().sort_index().to_dict())
from heart_disease.eda import plot_eda
X_train, y_train = splits["train"]
paths = plot_eda(X_train.assign(target=y_train), ROOT / "reports/figures")
from IPython.display import Image, display
for path in paths:
    display(Image(filename=str(path)))
```
